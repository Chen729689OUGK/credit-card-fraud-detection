## Credit Card Fraud Detection Project

```python
# Import the necessary modules

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
import itertools

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, recall_score, f1_score
```

```python
# Load the csv file

dataframe = pd.read_csv("creditcard.csv")
dataframe.head()
```

### Perform Exploratory Data Analysis

```python
dataframe.info()
```

```python
# Check for null values

dataframe.isnull().values.any()
```

```python
dataframe["Amount"].describe()
```

```python
non_fraud = len(dataframe[dataframe.Class == 0])
fraud = len(dataframe[dataframe.Class == 1])
fraud_percent = (fraud / (fraud + non_fraud)) * 100

print("Number of Genuine transactions: ", non_fraud)
print("Number of Fraud transactions: ", fraud)
print("Percentage of Fraud transactions: {:.4f}".format(fraud_percent))
```

```python
# Visualize the "Labels" column in our dataset

labels = ["Genuine", "Fraud"]
count_classes = dataframe.value_counts(dataframe['Class'], sort= True)
count_classes.plot(kind = "bar", rot = 0)
plt.title("Visualization of Labels")
plt.ylabel("Count")
plt.xticks(range(2), labels)
plt.show()
```

```python
# Perform Scaling
scaler = StandardScaler()
dataframe["NormalizedAmount"] = scaler.fit_transform(dataframe["Amount"].values.reshape(-1, 1))
dataframe.drop(["Amount", "Time"], inplace= True, axis= 1)

Y = dataframe["Class"]
X = dataframe.drop(["Class"], axis= 1)
```

```python
Y.head()
```

```python
# Split the data
(train_X, test_X, train_Y, test_Y) = train_test_split(X, Y, test_size= 0.3, random_state= 42)

print("Shape of train_X: ", train_X.shape)
print("Shape of test_X: ", test_X.shape)
```

Let's train different models on our dataset and observe which algorithm works better for our problem.

Let's apply Random Forests and Decision Trees algorithms to our dataset.

```python
# Decision Tree Classifier
decision_tree = DecisionTreeClassifier()
decision_tree.fit(train_X, train_Y)

predictions_dt = decision_tree.predict(test_X)
decision_tree_score = decision_tree.score(test_X, test_Y) * 100
```

```python
# Random Forest
random_forest = RandomForestClassifier(n_estimators= 100)
random_forest.fit(train_X, train_Y)

predictions_rf = random_forest.predict(test_X)
random_forest_score = random_forest.score(test_X, test_Y) * 100
```

```python
# Print scores of our classifiers

print("Random Forest Score: ", random_forest_score)
print("Decision Tree Score: ", decision_tree_score)
```

```python
# The below function is directly taken from the scikit-learn website to plot the confusion matrix

def plot_confusion_matrix(cm, classes, normalize=False, title='Confusion Matrix', cmap=plt.cm.Blues):
    """
    This function prints and plots the confusion matrix.
    Normalization can be applied by setting `normalize=True`.
    """
    if normalize:
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
        print("Normalized confusion matrix")
    else:
        print('Confusion matrix, without normalization')

    plt.imshow(cm, interpolation='nearest', cmap=cmap)
    plt.title(title)
    plt.colorbar()
    tick_marks = np.arange(len(classes))
    plt.xticks(tick_marks, classes, rotation=0)
    plt.yticks(tick_marks, classes)

    fmt = '.2f' if normalize else 'd'
    thresh = cm.max() / 2.
    for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
        plt.text(j, i, format(cm[i, j], fmt), horizontalalignment="center", color="white" if cm[i, j] > thresh else "black")

    plt.ylabel('True label')
    plt.xlabel('Predicted label')
    plt.tight_layout()

```

```python
# Plot confusion matrix for Decision Trees

confusion_matrix_dt = confusion_matrix(test_Y, predictions_dt.round())
print("Confusion Matrix - Decision Tree")
print(confusion_matrix_dt)
```

```python
plot_confusion_matrix(confusion_matrix_dt, classes=[0, 1], title= "Confusion Matrix - Decision Tree")
```

```python
# Plot confusion matrix for Random Forests

confusion_matrix_rf = confusion_matrix(test_Y, predictions_rf.round())
print("Confusion Matrix - Random Forest")
print(confusion_matrix_rf)
```

```python
plot_confusion_matrix(confusion_matrix_rf, classes=[0, 1], title= "Confusion Matrix - Random Forest")
```

```python
# The below function prints the following necesary metrics

def metrics(actuals, predictions):
    print("Accuracy: {:.5f}".format(accuracy_score(actuals, predictions)))
    print("Precision: {:.5f}".format(precision_score(actuals, predictions)))
    print("Recall: {:.5f}".format(recall_score(actuals, predictions)))
    print("F1-score: {:.5f}".format(f1_score(actuals, predictions)))
    
```

```python
print("Evaluation of Decision Tree Model")
print()
metrics(test_Y, predictions_dt.round())
```

```python
print("Evaluation of Random Forest Model")
print()
metrics(test_Y, predictions_rf.round())
```

Clearly, Random Forest model works better than Decision Trees

But, if we clearly observe our dataset suffers a serious problem of **class imbalance**. 
The genuine (not fraud) transactions are more than 99% with the fraud transactions constituting of 0.17%.

With such kind of distribution, if we train our model without taking care of the imbalance issues, it predicts the label with higher importance given to genuine transactions (as there are more data about them) and hence obtains more accuracy.

The class imbalance problem can be solved by various techniques. **Over sampling** is one of them.
 
One approach to addressing imbalanced datasets is to oversample the minority class. The simplest approach involves duplicating examples in the minority class, although these examples don’t add any new information to the model. 

Instead, new examples can be synthesized from the existing examples. This is a type of data augmentation for the minority class and is referred to as the **Synthetic Minority Oversampling Technique**, or **SMOTE** for short.

```python
# Performing oversampling on RF and DT

from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

# Step 1: Split the original dataset
train_X, test_X, train_Y, test_Y = train_test_split(
    X,
    Y,
    test_size=0.3,
    random_state=42,
    stratify=Y
)

# Step 2: Apply SMOTE only to the training data
smote = SMOTE(random_state=42)
X_resampled, Y_resampled = smote.fit_resample(train_X, train_Y)

print("Original training shape:", train_X.shape)
print("Resampled training shape:", X_resampled.shape)
print("Class distribution after SMOTE:")
print(Y_resampled.value_counts() if hasattr(Y_resampled, "value_counts") else __import__("collections").Counter(Y_resampled))

# Step 3: Train the Random Forest model
rf_resampled = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_resampled.fit(X_resampled, Y_resampled)

# Step 4: Predict on the untouched test set
predictions_resampled = rf_resampled.predict(test_X)

# Step 5: Evaluate the model
print("\nRandom Forest with SMOTE")
print("Accuracy:", accuracy_score(test_Y, predictions_resampled))
print("Precision:", precision_score(test_Y, predictions_resampled, zero_division=0))
print("Recall:", recall_score(test_Y, predictions_resampled, zero_division=0))
print("F1-score:", f1_score(test_Y, predictions_resampled, zero_division=0))

# Step 6: Display the confusion matrix
cm_resampled = confusion_matrix(test_Y, predictions_resampled)

print("\nConfusion Matrix - Random Forest with SMOTE")
print(cm_resampled)

plot_confusion_matrix(
    cm_resampled,
    classes=[0, 1],
    title="Confusion Matrix - Random Forest with SMOTE"
)
```

```python
# Build the Random Forest classifier on the new dataset

rf_resampled = RandomForestClassifier(n_estimators = 100)
rf_resampled.fit(train_X, train_Y)

predictions_resampled = rf_resampled.predict(test_X)
random_forest_score_resampled = rf_resampled.score(test_X, test_Y) * 100
```

```python
# Visualize the confusion matrix

cm_resampled = confusion_matrix(test_Y, predictions_resampled)
print("Confusion Matrix - Random Forest")
print(cm_resampled)
```

```python
plot_confusion_matrix(cm_resampled, classes=[0, 1], title= "Confusion Matrix - Random Forest After Oversampling")
```

```python
print("Evaluation of Random Forest Model with SMOTE")
print()
metrics(test_Y, predictions_resampled.round())
```

The Random Forest model was evaluated before and after applying SMOTE to the training data. The results are compared using precision, recall, F1-score, and the confusion matrix. Since fraud detection involves highly imbalanced data, accuracy alone is not sufficient to determine which model performs better.
